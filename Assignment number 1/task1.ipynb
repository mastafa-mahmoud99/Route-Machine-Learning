```python
x=input("please input a string that contains both letters and numbers")
print(x)
letters =[]
numbers =[]
for char in x :
    if char.isalpha():
        letters.append(char)
    elif char.isdigit():
        numbers.append(int(char))

word ="".join(letters)
total_sum =sum(numbers)
print (f'Word:"{word}",Sum,:"{total_sum}"')

```

```python
def valid_pass(password):
    if not (8<= len(password)<=16):
        return "password is not valid."
    special_chars = "@#$%^&*!?"
    has_upper_latter =False
    has_spacial_char =False
    for char in password:
        if  char.isupper():
            has_upper_latter= True
        if  char in special_chars :
            has_spacial_char= True
    if  has_upper_latter and has_spacial_char:
        return "The password is valid."
    else:
        return"The password is not valid."
z =input("Enter the password : ")
print (z)
result =valid_pass(z)
print (result) 

    
```

# Q3

```python
def final_price(price , is_member):
    if price<50:
        discount =0
    elif price<=100:
        discount =0.1
    else:
        discount =0.2
    final_price_after_discount =price*(1-discount)

    if is_member.lower() == "yes":
        final_price_after_discount =final_price_after_discount*0.95
    
    return final_price_after_discount

price = float(input("Enter total price: "))
is_member = input("Is the user a Member? (yes/no): ")

z= final_price(price,is_member)
print(f"Final price : {z}")

```

# Q9

```python
THE_number = int (input("enter the number : "))
factorial = 1 
for i in range (1,THE_number+1):
    factorial =factorial*i
#print (f"factorial = {factorial}")

sum =0
for i in range (1,THE_number+1):
    sum = sum + i
#print ("the sum  = ",  sum)

if factorial % sum == 0:
  print(
      f"Factorial is {factorial}, Sum is {sum}. {factorial} is divisible"
      f" by {sum}."
  )
else:
  print(
      f"Factorial is {factorial}, Sum is {sum}. {factorial} is not"
      f" divisible by {sum}."
  )
```

# Q10

```python
store ={
    "apple ":0.5,
    "bread": 2.0,
    "milk" :1.5,
}
cart =[]
while True:
    action = input("enter add or remove or done").strip().lower()
    if action == "done" :
        break
    item = input("enter name item ").strip().lower()
    if action == "add":
        if item in store:
            cart.append(item)
            print(f"added {item} to the cart")
        else:
            print ("item not in the store")
    elif action == "remove":
        if item in cart:
            cart.remove(item)
            print (f"removed {item} from cart")
        else:
            print("item is not in cart")
    else:
        print ("invalid action ")

total_bill =0
for item in cart :
    total_bill= total_bill +store[item]
print (f"total bill : {total_bill}")
```

```python
NUMBERS =[1,5,2,6,3]
peak =[]
for i in range(1, len(NUMBERS)-1):
    left = NUMBERS[i-1]
    current =NUMBERS[i]
    right = NUMBERS[i+1]

    if current>left and current> right :
        peak.append(current)
print (f"Peaks : {peak}")
```
